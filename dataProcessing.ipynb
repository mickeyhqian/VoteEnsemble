# Bike sharing data processing

```python
# fetch dataset 
bike_sharing = fetch_ucirepo(id=275) 
  
# data (as pandas dataframes) 
X: pd.DataFrame = bike_sharing.data.features 
y: pd.DataFrame = bike_sharing.data.targets 
  
# metadata 
print(bike_sharing.metadata) 
  
# variable information 
print(bike_sharing.variables)
```

```python
XComplete = pd.concat(
    [pd.get_dummies(X["mnth"], prefix="mnth", dtype=int), 
    pd.get_dummies(X["hr"], prefix="hr", dtype=int), 
    pd.get_dummies(X["weathersit"], prefix="weathersit", dtype=int), 
    X.loc[:, ["holiday", "workingday", "temp", "atemp", "hum", "windspeed"]]], axis=1
)
XComplete
```

```python
XComplete.iloc[:, -4:] = (XComplete.iloc[:, -4:] - XComplete.iloc[:, -4:].mean()) / XComplete.iloc[:, -4:].std()
```

```python
XComplete
```

```python
y
```

```python
yNorm = np.log(y+1)
yNorm = (yNorm - yNorm.mean()) / yNorm.std()
yNorm
```

```python
data = pd.concat([y, XComplete], axis=1)
data
```

```python
data.to_csv("/home/hqian/ResearchProjects/BaggingSAA/RealData/bike_sharing.csv", index=False)
```

# OpenML data

```python
import openml

# List all datasets and their properties
# openml.datasets.list_datasets(output_format="dataframe")

# Get dataset by ID
# dataset = openml.datasets.get_dataset(61)

# Get dataset by name
datasetFreq = openml.datasets.get_dataset('freMTPL2freq')
datasetSev = openml.datasets.get_dataset('freMTPL2sev')

# Get the data itself as a dataframe (or otherwise)
dataFreq = datasetFreq.get_data(dataset_format="dataframe")
dataSev = datasetSev.get_data(dataset_format="dataframe")
```

```python
print(len(dataFreq[0]["IDpol"]), len(set(dataFreq[0]["IDpol"])))
```

```python
IDToIndex = {}
for index in dataFreq[0].index:
    IDToIndex[dataFreq[0].loc[index, "IDpol"]] = index
```

```python
indexList = []
select = []
missingID = []
for index in dataSev[0].index:
    key = dataSev[0].loc[index, "IDpol"]
    if key in IDToIndex:
        indexList.append(IDToIndex[key])
        select.append(True)
    else:
        missingID.append(key)
        select.append(False)
```

```python
dataSev2 = pd.concat([dataSev[0].loc[select].reset_index(), dataFreq[0].loc[indexList].reset_index()], axis=1).drop(["index", "IDpol", "ClaimNb"], axis=1)
dataSev2
```

```python
numericFeatures = ["Exposure", "VehPower", "VehAge", "DrivAge", "BonusMalus", "Density"]

(dataSev2.loc[:, numericFeatures] - dataSev2.loc[:, numericFeatures].mean()) / dataSev2.loc[:, numericFeatures].std()
```

```python
XComplete = pd.concat(
    [pd.get_dummies(dataSev2["Area"], prefix="Area", dtype=int), 
    pd.get_dummies(dataSev2["VehBrand"], prefix="VehBrand", dtype=int), 
    pd.get_dummies(dataSev2["VehGas"], prefix="VehGas", dtype=int), 
    pd.get_dummies(dataSev2["Region"], prefix="Region", dtype=int), 
    (dataSev2.loc[:, numericFeatures] - dataSev2.loc[:, numericFeatures].mean()) / dataSev2.loc[:, numericFeatures].std()], axis=1
)
XComplete
```

```python
data = pd.concat([dataSev2["ClaimAmount"], XComplete], axis=1)
data
```

```python
data.to_csv("/home/hqian/ResearchProjects/VoteEnsemble/RealData/freMTPL2.csv", index=False)
```
