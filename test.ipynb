```python
import os 
import numpy as np

def dumpBagTail(testDir: str, sampleSize: int):
    veFile = os.path.join(testDir, "evalResults.pkl")
    bagFile = os.path.join(testDir, "1024_2048_4096_8192_16384_32768_65536.pkl")
    
    with open(veFile, "rb") as f:
        resultVE = pickle.load(f)
        
    with open(bagFile, "rb") as f:
        resultBag = pickle.load(f)
        
    qList = [0.5, 0.8, 0.9, 0.95, 0.975, 0.99]
    table = f"||50%|80%|90%|95%|97.5%|99%|\n"
    table += f"|-|-|-|-|-|-|-|\n"
    
    def createRow(method: str, quantiles: list[float]):
        row = f"|{method}"
        for value in quantiles:
            row += f"|{round(value, 3)}"
        return row + "|\n"
        
    for i in range(len(resultVE[8])):
        if resultVE[8][i] != sampleSize:
            continue
        quantiles = np.quantile(resultVE[0][i], q=qList)
        table += createRow("base", quantiles)
        quantiles = np.quantile(resultVE[2][i][0][0], q=qList)
        table += createRow("ROVE", quantiles)
        quantiles = np.quantile(resultVE[3][i][0][0], q=qList)
        table += createRow("ROVEs", quantiles)
        quantiles = np.quantile(list(resultBag[sampleSize].values()), q=qList)
        table += createRow("Bagging", quantiles)
    
    print(table)
    
    

def dumpVETail(testDir: str):
    veFile = os.path.join(testDir, "evalResults.pkl")
    
    with open(veFile, "rb") as f:
        resultVE = pickle.load(f)
        
    qList = [0.5, 0.8, 0.9, 0.95, 0.975, 0.99]
    table = f"||50%|80%|90%|95%|97.5%|99%|\n"
    table += f"|-|-|-|-|-|-|-|\n"
    
    def createRow(method: str, quantiles: list[float]):
        row = f"|{method}"
        for value in quantiles:
            row += f"|{round(value, 3)}"
        return row + "|\n"
        
    for i in range(len(resultVE[8])):
        quantiles = np.quantile(resultVE[0][i], q=qList)
        table += createRow("base", quantiles)
        quantiles = np.quantile(resultVE[2][i][0][0], q=qList)
        table += createRow("ROVE", quantiles)
        quantiles = np.quantile(resultVE[3][i][0][0], q=qList)
        table += createRow("ROVEs", quantiles)
    
    print(table)
        
```

```python
import json

def dumpTreeTail(VEDir: str, RFDir: str, GBDTDir: str, XGBDir: str, sampleSize: int):
    veFile = os.path.join(VEDir, "resultBrief.json")
    RFFile = os.path.join(RFDir, "resultBrief.json")
    GBDTFile = os.path.join(GBDTDir, "resultBrief.json")
    XGBFile = os.path.join(XGBDir, "resultBrief.json")
    
    with open(veFile, "r") as f:
        resultVE = json.load(f)
        
    with open(RFFile, "r") as f:
        resultRF = json.load(f)
        
    with open(GBDTFile, "r") as f:
        resultGBDT = json.load(f)
        
    with open(XGBFile, "r") as f:
        resultXGB = json.load(f)
        
    table = f"||50%|80%|90%|95%|97.5%|99%|\n"
    table += f"|-|-|-|-|-|-|-|\n"
    
    def getQuantiles(data: dict):
        return [data["50% quantile"], data["80% quantile"], data["90% quantile"], data["95% quantile"], data["97.5% quantile"], data["99% quantile"]]
    
    def createRow(method: str, quantiles: list[float]):
        row = f"|{method}"
        for value in quantiles:
            row += f"|{round(value, 3)}"
        return row + "|\n"
        
    table += createRow("base", getQuantiles(resultVE[f"sampleSize_{sampleSize}"]["base"]))
    table += createRow("ROVE", getQuantiles(resultVE[f"sampleSize_{sampleSize}"]["ROVE"]["(30, 200)"]["((30, 0.1), (30, 0.005))"]))
    table += createRow("ROVEs", getQuantiles(resultVE[f"sampleSize_{sampleSize}"]["ROVEs"]["(30, 200)"]["((30, 0.1), (30, 0.005))"]))
    table += createRow("RF", getQuantiles(resultRF[f"sampleSize_{sampleSize}"]["base"]))
    table += createRow("GBDT", getQuantiles(resultGBDT[f"sampleSize_{sampleSize}"]["base"]))
    table += createRow("XGB", getQuantiles(resultXGB[f"sampleSize_{sampleSize}"]["base"]))
    print(table)
```
